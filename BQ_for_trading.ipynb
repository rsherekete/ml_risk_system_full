```python
from google.cloud import bigquery
from google.colab import auth
import pandas as pd

# Authenticate the user
auth.authenticate_user()

# Initialize the BigQuery client with the project ID
project_id = 'zfx-dwh-prod'
client = bigquery.Client(project=project_id)

# Define the SQL query
query = """
SELECT *
FROM `zfx-dwh-prod.mt4_live01.orders`
WHERE TIMESTAMP_TRUNC(_close_ts_partition, DAY) = TIMESTAMP("2026-08-24")
LIMIT 1000
"""

# Execute the query and convert to a pandas DataFrame
df = client.query(query).to_dataframe()

# Display the first few rows
df.head()
```

```python
print(f"Datasets in project {project_id}:")

datasets = list(client.list_datasets())

if datasets:
    for dataset in datasets:
        if 'mt4' not in dataset.dataset_id:
            continue
        dataset_id = dataset.dataset_id
        print(f"\nDataset: {dataset_id}")

        # List tables in each dataset
        tables = list(client.list_tables(dataset_id))
        if tables:
            for table in tables:
                print(f"  - Table: {table.table_id}")
        else:
            print("  - No tables found in this dataset.")
else:
    print("No datasets found.")
```
